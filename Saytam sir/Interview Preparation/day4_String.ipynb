# Anagram

```python
def is_anagram(s1,s2):
    return sorted(s1) == sorted(s2)


if is_anagram("listen","silent"):
    print("Anagram")
else:
    print("Not Anagram")
```

```python
f = "Devesh"
freq = {}

for i in f:
    freq[i] = freq.get(i,0)+1

for i, c in freq.items():
    print(i,c)
```
