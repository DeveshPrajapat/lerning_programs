# Factorial (Loop)

```python
n = 5 
f = 1
for i in range(1,n+1):
    f*=i
print(f"Factorial of {f}")
    
```

# Factorial (Recursion)

```python
def fact(n):
    if n == 1:
        return 1
    return n*fact(n-1)

print(f"Factorial is {fact(5)}")
```

# Fibonacci (Loop)

```python
n = 5
a = 0 
b = 1
for i in range(n+1):
    print(a,end=" ")
    a,b = b,a+b
```

# Fibonacci (Recusion)

```python
def fibo(n):
    if n <= 1:
        return n
    return fibo(n-1) + fibo(n-2)

for i in range(5):
    print(fibo(i),end=" ")
```
